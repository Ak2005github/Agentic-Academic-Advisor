## 3. Test advisor graph

Without API keys, the graph still routes and returns evidence summaries. With `OPENAI_API_KEY` or `OPEN_AI_KEY`, it will generate a cleaner natural-language answer. With Pinecone configured, RAG will include dense retrieval too.

```python
from modules.advisor_graph import run_advisor

questions = [
    "What are the prerequisites for CS402?",
    "Can student 1001 take CS402 next semester?",
    "How many credits does student 1001 still need for the degree?",
    "Can student 1001 take CS402 and what policy applies to prerequisite overrides?",
]

for question in questions:
    result = run_advisor(question)
    print("QUESTION:", question)
    print("ROUTES:", result.get("routes"))
    print("ANSWER:", result.get("answer"))
    print("-" * 80)
```

```python
questions2 = ["What are the undergraduate admission requirements?",
              "What is the admission form submission process?",
              "What documents are required for undergraduate admission?",
              "What is the application photograph requirement?",
              "What happens after passing the pre-admission entry test?"
             ]

for question in questions2:
    result = run_advisor(question)
    print("QUESTION:", question)
    print("ROUTES:", result.get("routes"))
    print("ANSWER:", result.get("answer"))
    print("-" * 80)
```

```python
for question in questions2:
    result = run_advisor(question)
    print("QUESTION:", question)
    print("ROUTES:", result.get("routes"))
    print("ANSWER:", result.get("answer"))

    if result.get("llm_error"):
        print("LLM ERROR:", result.get("llm_error"))

    print("-" * 80)
```
